```python
# imports

import pandas as pd
import os
```

```python
# load the scored reviews
df = pd.read_csv('../data/processed/rmp_scored.csv')

# only reviews with text
df = df[df['comment'].notna()]

# ----- LLM sample -----

# 500 random reviews from each school
llm_sample = df.groupby('school').sample(n=500, random_state=42)

print('LLM sample size', len(llm_sample))
print(llm_sample['school'].value_counts())

# ----- human sample -----

# 25 reviews per school, taken from inside the LLM sample
human_sample = llm_sample.groupby('school').sample(n=25, random_state=42)

print('Human sample size', len(human_sample))

# ----- save the samples -----
os.makedirs('../data/samples', exist_ok=True)
llm_sample.to_csv('../data/samples/llm_sample.csv', index=False)
human_sample.to_csv('../data/samples/human_sample.csv', index=False)

# ----- labeling sheets -----

label_sheet = human_sample[['review_id', 'comment']].sample(frac=1, random_state=42)

# empty columns for each person to fill in
label_sheet['leniency'] = ''
label_sheet['teaching'] = ''
label_sheet['grade_complaint'] = ''

# save the empty file for filling out
os.makedirs('../data/labels', exist_ok=True)
label_sheet.to_excel('../data/labels/labeling_sheet.xlsx', index=False)
```
